```python
from datasets import load_dataset

# Örn. basit bir text dataset
dataset = load_dataset("ag_news", split="train[:1000]")  # ilk 1000 satır
texts = dataset["text"]
```

```python
from sentence_transformers import SentenceTransformer

# Daha güncel, hızlı model
model = SentenceTransformer("sentence-transformers/all-MiniLM-L12-v2")
embeddings = model.encode(texts, batch_size=64, convert_to_numpy=True, show_progress_bar=True)
```

```python
from langchain.vectorstores import FAISS
from langchain.embeddings import HuggingFaceEmbeddings

embedder = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L12-v2")
docsearch = FAISS.from_texts(texts, embedder)

query = "economy news about stock market"
docs = docsearch.similarity_search(query, k=3)

for d in docs:
    print(d.page_content)
```
